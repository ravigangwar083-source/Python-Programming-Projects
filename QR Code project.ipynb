```python
import qrcode
url=input("Enter url:")
if url.startswith("https://") or url.startswith("http://"):
    # version can be given in the range of 1 to 40
    # It defines the starting size of the qrcode i,e setd the level of density.
    # If you take version 1  it will make 21 * 21 sq. pixel QR code.
     # H- High M- Medium Desnsity, L- Low density
    qr=qrcode.QRCode(version=1,
                    error_correction=qrcode.constants.ERROR_CORRECT_H,
                    box_size=10, #10*10
                    border=4
                    )
    qr.add_data(url) #add_adata- add data according to the given abaove version of qr code 
    qr.make()
    image=qr.make_image(fill_color="blue", back_color="red")
    image.save("Third_QR.png")
else:
    Print("invalid url")
```

```python
#create QR Code for UPI
import qrcode

upi_id=input("Enter UPI ID:")
name=input("Enter Name:")

upi_url=f"upi://pay?pa={upi_id}&pn={name}"

qr=qrcode.make(upi_url)
qr.save("Ravi_UPI2.png")
print("UPI ka QR ban Gaya")
```

```python
#create QR Code for UPI with specific amount
import qrcode

upi_id=input("Enter UPI ID:")
name=input("Enter Name:")
amount=int(input("Enter Amount:"))

upi_url=f"upi://pay?pa={upi_id}&pn={name}&am={amount}"

qr=qrcode.make(upi_url)
qr.save("Specific amount.png")
print("UPI ka QR ban Gaya")
```
